```python
##Part B — Missing Data & Cleaning
```

```python
#1. Convert a numeric-looking text column using pd.to_numeric().
```

```python
df['Quantity']=pd.to_numeric(df['Quantity'],errors='coerce')
df['UnitPrice']=pd.to_numeric(df['UnitPrice'],errors='coerce')
df['Debit']=pd.to_numeric(df['Debit'],errors='coerce')
```

```python
#2. Use fillna() to replace missing numeric values with 0.
```

```python
numeric_cols=df.select_dtypes(include='number').columns
df[numeric_cols] = df[numeric_cols].fillna(0)
```

```python
df.isna().sum()
```

```python
 #3. Fill missing UnitPrice values using the mean UnitPrice.
```

```python
mean_unitprice=df['UnitPrice'].mean()
df['UnitPrice']=df['UnitPrice'].replace({0:mean_unitprice})
df['UnitPrice'].value_counts()
```

```python
#4. Use dropna() and compare row counts before and after.
```

```python
before = len(df)
print("Rows before dropna:", before)
```

```python
df_clean = df.dropna()
after = len(df_clean)

print("Rows after dropna:", after)
```

```python
print("Rows removed:", before - after)
```

```python
#5. Find rows where Invoice No is missing.
```

```python
df[df["Invoice No"].isna()]
```

```python
#6. Remove leading/trailing spaces from Supplier names.
```

```python
df['Supplier']=df['Supplier'].str.title()
df['Supplier']=df['Supplier'].str.strip()

df['Supplier'].head()
```

```python
#7. Convert a text column to lowercase using .str.
```

```python
df['Description']=df['Description'].str.strip().str.lower()
df['Description'].head()
```

```python
#10.Correct and filling the Date Column into standard format
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

mode_date = df['Date'].mode()[0]

df['Date'] = df['Date'].fillna(mode_date)

print("Date column after conversion and filling missing values:")
display(df.info())
display(df.head())
```

```python
##Part C — Invoice & Transaction Validation
```

```python
df['transaction_value']=df['Quantity']*df['UnitPrice']
df['transaction_value']
```

```python
#Part F — Practical Finance Challenges
```

```python
#1. Create an invoice exception report containing duplicates, missing values and validation failures.
```

```python
invoice_exception_report=pd.DataFrame({"Invoice Duplicates":df.loc[df['Invoice No'].duplicated(keep=False),'Invoice No'],
                                       "Missing Values Count":df.isnull().sum(axis=1),
                                       "Validation Failures":df['validation']

})
invoice_exception_report
```

```python
#2. Create a supplier summary with transaction count, total Debit, total Credit and net amount.
```

```python
supplier_summary = df.groupby('Supplier').agg(
    Transaction_Count=('Supplier', 'size'),
    Total_Debit=('Debit', 'sum'),
    Total_Credit=('Credit', 'sum')
)
supplier_summary['Net Amount'] = supplier_summary['Total_Debit'] - supplier_summary['Total_Credit']
supplier_summary
```
